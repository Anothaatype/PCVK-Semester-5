## **TASK 1 : IMAGE INVERSION**

```python
def task1_image_inversion():
    """
    Task 1: Implement image inversion
    Formula: inverted_pixel = 255 - original_pixel
    """
    print("=== TASK 1: IMAGE INVERSION ===")

    # Load the image
    img = cv2.imread(image_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Apply inversion formula: 255 - pixel_value
    inverted_img = 255 - img_rgb

    # Display results
    plt.figure(figsize=(15, 6))

    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title('Original Image', fontsize=14)
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(inverted_img)
    plt.title('Inverted Image\nFormula: 255 - pixel_value', fontsize=14)
    plt.axis('off')

    plt.suptitle('Task 1: Image Inversion Transformation', fontsize=16)
    plt.tight_layout()
    plt.show()

    print("✓ Task 1: Image inversion completed successfully!")
    return img_rgb, inverted_img

# Run Task 1
original, inverted = task1_image_inversion()
```

## **TASK 2 - CONTRAST TRANSFORMATION**

```python
def task2_contrast_transformation(brightness_range=50, contrast_factor=2):
    """
    Task 2: Implement contrast transformation
    Parameters as shown in assignment:
    - brightness_range [-255 - 255]: 50
    - contrast_factor [1.0 - 3.0]: 2
    """
    print("=== TASK 2: CONTRAST TRANSFORMATION ===")
    print("Mengubah kontras dan tingkat kecerahan citra")
    print("-" * 45)
    print(f"Masukkan tingkat kecerahan [-255 - 255]: {brightness_range}")
    print(f"Masukkan kontras [1.0 - 3.0]: {contrast_factor}")

    # Load the image
    img = cv2.imread(image_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Apply contrast transformation
    # Formula: new_pixel = alpha * pixel + beta
    # where alpha = contrast_factor, beta = brightness_range
    contrast_img = cv2.convertScaleAbs(img_rgb, alpha=contrast_factor, beta=brightness_range)

    # Display results
    plt.figure(figsize=(15, 6))

    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title('Original Image', fontsize=14)
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(contrast_img)
    plt.title(f'Contrast Enhanced\nBrightness: {brightness_range}, Contrast: {contrast_factor}', fontsize=14)
    plt.axis('off')

    plt.suptitle('Task 2: Contrast Transformation', fontsize=16)
    plt.tight_layout()
    plt.show()

    print("✓ Task 2: Contrast transformation completed successfully!")
    return img_rgb, contrast_img

# Run Task 2
original, contrast_enhanced = task2_contrast_transformation()
```

## **TASK 3 - LOGARITHMIC TRANSFORMATION**

```python
def task3_logarithmic_transformation(brightness_value=50):
    """
    Task 3: Implement logarithmic brightness transformation
    Formula: new_pixel = c * log(1 + pixel)
    """
    print("=== TASK 3: LOGARITHMIC BRIGHTNESS TRANSFORMATION ===")
    print("Mengubah tingkat kecerahan citra dengan Transformasi Log")
    print("-" * 55)
    print(f"Masukkan nilai kecerahan: {brightness_value}")

    # Load the image
    img = cv2.imread(image_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Convert to float for calculation
    img_float = img_rgb.astype(np.float64)

    # Apply logarithmic transformation
    # Formula: s = c * log(1 + r)
    c = brightness_value / np.log(1 + np.max(img_float))
    log_img = c * np.log(1 + img_float)

    # Convert back to uint8
    log_img = np.uint8(np.clip(log_img, 0, 255))

    # Display results
    plt.figure(figsize=(15, 6))

    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title('Original Image', fontsize=14)
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(log_img)
    plt.title(f'Log Transformed\nFormula: c * log(1 + pixel)\nc = {c:.2f}', fontsize=14)
    plt.axis('off')

    plt.suptitle('Task 3: Logarithmic Brightness Transformation', fontsize=16)
    plt.tight_layout()
    plt.show()

    print("✓ Task 3: Logarithmic transformation completed successfully!")
    return img_rgb, log_img

# Run Task 3
original, log_transformed = task3_logarithmic_transformation()
```

## **TASK 4 - GRAYSCALE TRANSFORMATIONS :**

```python
def task4_grayscale_transformations():
    """
    Task 4: Implement three methods of grayscale conversion:
    a. Averaging: Gray = (R + G + B) / 3
    b. Lightness: Gray = (max(R,G,B) + min(R,G,B)) / 2
    c. Luminance: Gray = 0.299*R + 0.587*G + 0.114*B
    """
    print("=== TASK 4: GRAYSCALE TRANSFORMATIONS ===")

    # Load the image
    img = cv2.imread(image_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Method A: Averaging
    gray_avg = np.mean(img_rgb, axis=2).astype(np.uint8)
    print("✓ Method A - Averaging completed")

    # Method B: Lightness
    gray_lightness = ((np.max(img_rgb, axis=2) + np.min(img_rgb, axis=2)) / 2).astype(np.uint8)
    print("✓ Method B - Lightness completed")

    # Method C: Luminance (Weighted average)
    gray_luminance = (0.299 * img_rgb[:,:,0] + 0.587 * img_rgb[:,:,1] + 0.114 * img_rgb[:,:,2]).astype(np.uint8)
    print("✓ Method C - Luminance completed")

    # Display results
    plt.figure(figsize=(20, 12))

    # Original image
    plt.subplot(2, 4, 1)
    plt.imshow(img_rgb)
    plt.title('Original Image', fontsize=14)
    plt.axis('off')

    # Empty space
    plt.subplot(2, 4, 2)
    plt.axis('off')

    # Empty space
    plt.subplot(2, 4, 3)
    plt.axis('off')

    # Empty space
    plt.subplot(2, 4, 4)
    plt.axis('off')

    # Method A: Averaging
    plt.subplot(2, 4, 5)
    plt.imshow(gray_avg, cmap='gray')
    plt.title('a. Averaging\nGray = (R + G + B) / 3', fontsize=12)
    plt.axis('off')

    # Method B: Lightness
    plt.subplot(2, 4, 6)
    plt.imshow(gray_lightness, cmap='gray')
    plt.title('b. Lightness\nGray = (max + min) / 2', fontsize=12)
    plt.axis('off')

    # Method C: Luminance
    plt.subplot(2, 4, 7)
    plt.imshow(gray_luminance, cmap='gray')
    plt.title('c. Luminance\nGray = 0.299R + 0.587G + 0.114B', fontsize=12)
    plt.axis('off')

    plt.suptitle('Task 4: Grayscale Transformation Methods', fontsize=16)
    plt.tight_layout()
    plt.show()

    print("✓ Task 4: All grayscale transformations completed successfully!")
    return img_rgb, gray_avg, gray_lightness, gray_luminance

# Run Task 4
original, gray_avg, gray_lightness, gray_luminance = task4_grayscale_transformations()
```

## **TASK 5 - SELECTIVE COLOR DISPLAY**

```python
def task5_selective_color_display(target_color='blue'):
    """
    Task 5: Display certain colors and convert other colors to grayscale
    Example: display blue and convert non-blue parts to grayscale
    """
    print(f"=== TASK 5: SELECTIVE COLOR DISPLAY ({target_color.upper()}) ===")

    # Load the image
    img = cv2.imread(image_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Convert to HSV for better color detection
    img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

    # Define color ranges in HSV
    color_ranges = {
        'blue': {
            'lower': np.array([100, 50, 50]),
            'upper': np.array([130, 255, 255])
        },
        'red': {
            'lower': np.array([0, 50, 50]),
            'upper': np.array([10, 255, 255])
        },
        'green': {
            'lower': np.array([40, 50, 50]),
            'upper': np.array([80, 255, 255])
        }
    }

    # Create mask for the target color
    if target_color in color_ranges:
        mask = cv2.inRange(img_hsv, color_ranges[target_color]['lower'], color_ranges[target_color]['upper'])
    else:
        # Default to blue if color not found
        mask = cv2.inRange(img_hsv, color_ranges['blue']['lower'], color_ranges['blue']['upper'])

    # Create grayscale version
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    gray_3channel = cv2.cvtColor(gray, cv2.COLOR_GRAY2RGB)

    # Create selective color image
    selective_img = gray_3channel.copy()
    selective_img[mask > 0] = img_rgb[mask > 0]

    # Display results
    plt.figure(figsize=(15, 6))

    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title('Original Image', fontsize=14)
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(selective_img)
    plt.title(f'Selective Color Display\n({target_color.capitalize()} preserved, others grayscale)', fontsize=14)
    plt.axis('off')

    plt.suptitle('Task 5: Selective Color Display', fontsize=16)
    plt.tight_layout()
    plt.show()

    print(f"✓ Task 5: Selective color display for {target_color} completed successfully!")
    return img_rgb, selective_img, mask

# Run Task 5
original, selective_blue, mask = task5_selective_color_display('blue')
```

## **TASK 1 - GAMMA CORRECTION**

```python
def task1_gamma_correction(gamma_value=3):
    """
    Task 1: Apply Gamma Correction
    Formula: output = (pixel/255)^(1/gamma) * 255
    """
    print("=== TASK 1: GAMMA CORRECTION ===")

    # Load image
    img = cv2.imread(image_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Build LUT
    invGamma = 1.0 / gamma_value
    table = np.array([((i / 255.0) ** invGamma) * 255
                      for i in np.arange(256)]).astype("uint8")
    gamma_corrected = cv2.LUT(img_rgb, table)

    # Display
    plt.figure(figsize=(15, 6))
    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title("Original Image", fontsize=14)
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(gamma_corrected)
    plt.title(f"Gamma Corrected (γ={gamma_value})", fontsize=14)
    plt.axis("off")

    plt.suptitle("Task 1: Gamma Correction", fontsize=16)
    plt.tight_layout()
    plt.show()

    print("✓ Task 1: Gamma correction completed!")
    return img_rgb, gamma_corrected

# Run Task 1
original, gamma_img = task1_gamma_correction(gamma_value=3)
```

## **TASK 2: Image Quantization**

```python
def task2_image_depth_simulation(bit_depth=2):
    """
    Task 2: Simulate reduced bit depth (quantization)
    """
    print("\n=== TASK 2: IMAGE QUANTIZATION ===")

    img = cv2.imread(image_path)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    levels = 2 ** bit_depth
    factor = 256 // levels
    quantized = (gray // factor) * factor

    plt.figure(figsize=(15, 6))
    plt.subplot(1, 2, 1)
    plt.imshow(gray, cmap="gray")
    plt.title("Original Grayscale", fontsize=14)
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(quantized, cmap="gray")
    plt.title(f"Quantized ({bit_depth} bits)", fontsize=14)
    plt.axis("off")

    plt.suptitle("Task 2: Image Quantization", fontsize=16)
    plt.tight_layout()
    plt.show()

    print("✓ Task 2: Image quantization completed!")
    return gray, quantized

# Run Task 2
gray_img, quantized_img = task2_image_depth_simulation(bit_depth=3)
```

## **TASK 3: Average Denoising**

```python
# ==== Task 3: Average Denoising Practicum ====

import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import glob, os
from skimage.metrics import peak_signal_noise_ratio
from google.colab import drive

# 1. Mount Google Drive
drive.mount('/content/drive', force_remount=True)

# 2. Define paths (your given paths)
original_path = '/content/drive/MyDrive/PCVK_MM/Images-Polinema/galaxy.jpg'
noisy_pattern = '/content/drive/MyDrive/PCVK_MM/Images-Polinema/noises/*.jpg'

# 3. Load original image (BGR → RGB)
original_bgr = cv.imread(original_path)
if original_bgr is None:
    raise FileNotFoundError(f"Original image not found at {original_path}")
original = cv.cvtColor(original_bgr, cv.COLOR_BGR2RGB)
h, w = original.shape[:2]
print("✓ Loaded original:", original.shape)

# 4. Load noisy images
cv_img = []
for img in sorted(glob.glob(noisy_pattern)):
    n = cv.imread(img)
    if n is None:
        print(f"Warning: could not read {img}, skipping")
        continue
    n = cv.resize(n, (w, h))  # resize if not same size
    cv_img.append(cv.cvtColor(n, cv.COLOR_BGR2RGB))

print(f"✓ Loaded {len(cv_img)} noisy images")

# 5. Average denoising function
def average_denoising(images_list, num_images):
    """Compute average denoising of first num_images (RGB uint8)."""
    selected = images_list[:num_images]
    acc = np.zeros_like(selected[0], dtype=np.float64)
    for img in selected:
        acc += img.astype(np.float64)
    avg = (acc / num_images).astype(np.uint8)
    return avg

# 6. PSNR function
def calculate_psnr(orig, denoised):
    return peak_signal_noise_ratio(orig, denoised, data_range=255)

# 7. Test with different numbers of images
test_numbers = [10, 20, 40, 80, 100]
results = []

rows = len(test_numbers)
fig, axes = plt.subplots(rows, 3, figsize=(14, 4*rows))

for idx, num in enumerate(test_numbers):
    if num > len(cv_img):
        print(f"⚠️ Only {len(cv_img)} noisy images available. Skipping {num}.")
        continue

    denoised = average_denoising(cv_img, num)
    psnr_val = calculate_psnr(original, denoised)
    results.append({'NumImages': num, 'PSNR_dB': psnr_val, 'Denoised': denoised})

    # Column 1: label
    axes[idx, 0].axis('off')
    axes[idx, 0].text(0.5, 0.5, f"{num} Images Average", ha='center', va='center', fontsize=12)
    axes[idx, 0].set_title("Number of Images")

    # Column 2: denoised image
    axes[idx, 1].imshow(denoised)
    axes[idx, 1].axis('off')
    axes[idx, 1].set_title("Denoised Result")

    # Column 3: PSNR value
    axes[idx, 2].axis('off')
    axes[idx, 2].text(0.5, 0.5, f"{psnr_val:.2f} dB", ha='center', va='center', fontsize=12)
    axes[idx, 2].set_title("PSNR")

plt.suptitle("Task 3: Average Denoising Results (Num | Image | PSNR)", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()

# 8. Show results as table
df = pd.DataFrame([{'NumImages': r['NumImages'], 'PSNR_dB': r['PSNR_dB']} for r in results])
print("\nRESULTS TABLE:")
display(df)
```

### ***PNSR Conslusion***

From the PSNR results, it can be concluded that the quality of the denoised image improves as the number of averaged noisy images increases. Starting from 10 averaged images with a PSNR of around 19.74 dB, the value gradually increases to 19.84 dB with 20 images, 19.89 dB with 40 images, and 19.92 dB with 80 images. At 100 images, the PSNR remains nearly the same at 19.92 dB, showing that the improvement becomes less significant as more images are added. This trend indicates that averaging is effective in reducing Gaussian noise, but after a certain number of images, the noise is already sufficiently minimized and the benefit of adding more averages becomes marginal.

## **Task 4: Image Masking with Logical Operators**

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive

# 1. Mount Google Drive
drive.mount('/content/drive', force_remount=True)

# 2. Load original image
image_path = '/content/drive/MyDrive/PCVK_MM/Images-Polinema/couple.tiff'
img_bgr = cv.imread(image_path)
if img_bgr is None:
    raise FileNotFoundError(f"Image not found at {image_path}")
img = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)

print("✓ Loaded image:", img.shape)

# 3. Create binary mask (circle)
mask = np.zeros(img.shape[:2], dtype=np.uint8)
h, w = mask.shape
center = (w//2, h//2)
radius = min(h, w) // 4
cv.circle(mask, center, radius, 255, -1)
mask_rgb = cv.merge([mask, mask, mask])

# 4. Apply operators
results = [
    ("NOT", img, cv.bitwise_not(img)),
    ("AND", mask_rgb, cv.bitwise_and(img, mask_rgb)),
    ("OR", mask_rgb, cv.bitwise_or(img, mask_rgb)),
    ("XOR", mask_rgb, cv.bitwise_xor(img, mask_rgb)),
    ("NAND", mask_rgb, cv.bitwise_not(cv.bitwise_and(img, mask_rgb)))
]

# 5. Plot results as table (Operator | Input | Output)
rows = len(results)
fig, axes = plt.subplots(rows, 3, figsize=(15, 3*rows))

for i, (op, input_img, output_img) in enumerate(results):
    # Column 1: Operator name
    axes[i, 0].axis("off")
    axes[i, 0].text(0.5, 0.5, op, ha="center", va="center", fontsize=12, weight="bold")
    axes[i, 0].set_title("Operator")

    # Column 2: Input image
    axes[i, 1].imshow(input_img, cmap='gray' if op=="NOT" else None)
    axes[i, 1].axis("off")
    axes[i, 1].set_title("Input Image")

    # Column 3: Output image
    axes[i, 2].imshow(output_img)
    axes[i, 2].axis("off")
    axes[i, 2].set_title("Output Image")

plt.suptitle("Task 4: Image Masking with Logical Operators", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

### ***Analysis Results***

From the results of the masking operations, it can be observed that each logical operator produces a different visual effect depending on how the image interacts with the binary mask. The AND operator keeps only the region inside the circular mask, effectively isolating part of the image while setting the outside to black. In contrast, the OR operator highlights the mask area by merging the mask with the image, making the circle region appear brighter. The XOR operator emphasizes the differences, so only the non-overlapping regions between the image and mask are shown, resulting in a boundary-like effect. The NOT operator inverts all pixel values, turning light regions dark and dark regions light, which completely reverses the original image. Finally, the NAND operator produces the inverse of the AND result, showing the masked circle region inverted while leaving the outside unchanged. These outcomes demonstrate how different bitwise operators control the interaction between images and masks, allowing selective enhancement, isolation, or inversion of regions based on binary logic.
